## Módulo 2: Variables, Tipos y Operaciones

**2.1 Conversión segura a float**

Implementa `to_float_seguro(s)` que reciba un string y retorne `float` manejando estos casos:
- Cambia coma decimal por punto si aparece (ej. `'3,14' -> 3.14`).
- Si no se puede convertir, retorna `None` en lugar de lanzar excepción.
Agrega al menos 4 pruebas rápidas.

```python
def to_float_seguro(s):
    # intento de convertir a float
    try:
        if s is None:
            return None
        # reemplazo coma por punto
        s = s.replace(",", ".")
        return float(s)
    except:
        # si no se puede, devuelvo None
        return None


# pruebas
print(to_float_seguro("3,14"))   # debería dar 3.14
print(to_float_seguro("2.5"))    # 2.5
print(to_float_seguro("hola"))   # None
print(to_float_seguro(""))       # None
print(to_float_seguro("1000"))   # 1000.0
```

"Hice una función que cambia la coma por punto y trata de pasar el string a float.  
Si falla, simplemente devuelve `None` en lugar de explotar.  
Probé con varios casos para ver que funciona. " 

## Módulo 3: Estructuras de Control

**3.1 Menú simulado con `while`**

Simula un flujo de opciones con una lista `entradas = ['1','2','x','q']`. Recorre esa lista con `while` y:
- Si la opción es `'1'`, imprime `hola`.
- Si es `'2'`, imprime el resultado de `2+2`.
- Si es otro valor distinto de `'q'`, imprime `opción inválida`.
- Finaliza cuando es `'q'`.
No uses `input()` real.

```python
entradas = ['1','2','x','q']

i = 0
while i < len(entradas):
    opcion = entradas[i]

    if opcion == '1':
        print("hola")
    elif opcion == '2':
        print(2+2)
    elif opcion == 'q':
        print("saliendo...")
        break
    else:
        print("opción inválida")

    i += 1
```

"Uso un `while` para recorrer la lista de opciones.  
Según el valor imprime un resultado, y si aparece `'q'` el programa se detiene."

## Módulo 4: Estructuras de Datos

**4.1 Diccionario de frecuencias (top 5)**

Dado `texto = 'Abracadabra, foco y acción'`, construye un **diccionario de frecuencias** de caracteres (ignora espacios) y muestra una lista con los **5** caracteres más frecuentes y su conteo, ordenados de mayor a menor.

```python
texto = 'Abracadabra, foco y acción'

# quitamos espacios
texto = texto.replace(" ", "")

# diccionario de frecuencias
frecuencias = {}
for c in texto:
    frecuencias[c] = frecuencias.get(c, 0) + 1

# ordenar por frecuencia (mayor a menor)
ordenados = sorted(frecuencias.items(), key=lambda x: x[1], reverse=True)

# top 5
top5 = ordenados[:5]

print("Diccionario completo:", frecuencias)
print("Top 5:", top5)
```

Se construyó un diccionario que cuenta cuántas veces aparece cada carácter,  
ignorando espacios. Luego se ordenó por frecuencia y se extrajeron los 5 más comunes.  
El resultado muestra tanto el diccionario completo como el top 5.  

## Módulo 5: Funciones y Módulos

**5.1 `*args` y parámetro con nombre**

Implementa `acumula(*args, operacion='suma')` que:
- Si `operacion='suma'`, retorna la suma de todos los `args` numéricos.
- Si `operacion='resta'`, retorna el primer valor menos el resto.
Incluye al menos 3 pruebas cortas (casos normales y un borde).

```python
def acumula(*args, operacion='suma'):
    if not args:   # si no hay números
        return 0
    
    if operacion == 'suma':
        total = 0
        for n in args:
            total += n
        return total
    elif operacion == 'resta':
        total = args[0]
        for n in args[1:]:
            total -= n
        return total
    else:
        return None


# pruebas
print(acumula(1, 2, 3))                  # suma normal -> 6
print(acumula(10, 2, 3, operacion='resta'))  # resta -> 5
print(acumula(5, operacion='resta'))     # borde, solo un número -> 5
print(acumula())                         # borde, vacío -> 0
```

## Módulo 6: Programación Orientada a Objetos (POO)

**6.1 Clase mínima con `__init__` y `__str__`**

Define una clase `Producto` con atributos `nombre` y `precio`. Implementa `__str__` para que al imprimir una instancia se vea exactamente `Producto(nombre, precio)`. Crea dos instancias y muéstralas.

```python
class Producto:
    def __init__(self, nombre, precio):
        self.nombre = nombre
        self.precio = precio
    
    def __str__(self):
        return f"Producto({self.nombre}, {self.precio})"


# crear instancias
p1 = Producto("Laptop", 1200)
p2 = Producto("Mouse", 25)

print(p1)
print(p2)
```

## Módulo 7: Archivos y Automatización

**7.1 JSON seguro con fallback de versión**

Crea un diccionario de configuración y guárdalo en `config.json` con `indent=2` y UTF-8. Luego, **cárgalo** y verifica que exista la clave `'version'`. Si falta, añade `'version': '1.0'` y vuelve a guardar. Muestra el JSON final por pantalla.

```python
import json

# diccionario de configuración inicial
config = {
    "app": "OeconomyApp",
    "autor": "Angel",
    "debug": True
}

# guardar en config.json
with open("config.json", "w", encoding="utf-8") as f:
    json.dump(config, f, indent=2, ensure_ascii=False)

# cargar de nuevo
with open("config.json", "r", encoding="utf-8") as f:
    datos = json.load(f)

# verificar clave 'version'
if "version" not in datos:
    datos["version"] = "1.0"

# volver a guardar
with open("config.json", "w", encoding="utf-8") as f:
    json.dump(datos, f, indent=2, ensure_ascii=False)

# mostrar json final
print(json.dumps(datos, indent=2, ensure_ascii=False))
```

## Módulo 8: Pandas y Matplotlib

**8.1 Ventas por canal con gráfico**

Construye un `DataFrame` con columnas `canal` y `ventas` (al menos 5 filas con 2–3 canales). Calcula ventas totales por canal con `groupby` y genera un gráfico de barras con Matplotlib (etiquetas de ejes y título).

```python
%pip install -q pandas matplotlib
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# datos simples (mín. 5 filas, 2–3 canales)
data = {
    "canal": ["Instagram", "Tienda", "Amazon", "Instagram", "Amazon", "Tienda", "Amazon"],
    "ventas": [1200, 800, 950, 1500, 600, 1100, 900]
}

df = pd.DataFrame(data)

# totales por canal con groupby
totales = (
    df.groupby("canal", as_index=False)["ventas"]
      .sum()
      .sort_values("ventas", ascending=False)
)

print("Datos originales:")
print(df, "\n")
print("Ventas totales por canal:")
print(totales)

# gráfico de barras (sin estilos raros)
plt.figure()
plt.bar(totales["canal"], totales["ventas"])
plt.xlabel("Canal")
plt.ylabel("Ventas totales")
plt.title("Ventas por canal")

# etiquetas encima de cada barra para que se vea el número
for i, v in enumerate(totales["ventas"]):
    plt.text(i, v, str(v), ha="center", va="bottom")

plt.show()
```

En este caso estadistico segun lo observado el canal mas eficiente de ventas para la empresa es instagram. El proceso es el siguiente: Se crea un `DataFrame` con columnas **canal** y **ventas** (7 filas, 3 canales).  
Con `groupby` se calculan las ventas totales por canal y se ordenan de mayor a menor.  
Luego se genera un **gráfico de barras** con Matplotlib con etiquetas de ejes y título.

## Módulo 9: Automatización Avanzada

**9.1 Comprimir `.txt` con `zipfile`**

Crea (si hace falta) 2–3 archivos `.txt` pequeños en el directorio actual y comprímelos en `textos.zip` usando `zipfile`. Al final, **lista** el contenido del zip para verificar.

```python
import zipfile
import os

# Crear algunos archivos .txt si no existen
archivos = ["Informe de ventas.txt", "Gastos.txt", "Beneficios.txt"]

for i, nombre in enumerate(archivos, start=1):
    if not os.path.exists(nombre):
        with open(nombre, "w", encoding="utf-8") as f:
            f.write(f"Este es el archivo {nombre}, número {i}\n")

# Comprimirlos en textos.zip
with zipfile.ZipFile("textos.zip", "w") as z:
    for nombre in archivos:
        z.write(nombre)

# Listar contenido del zip
with zipfile.ZipFile("textos.zip", "r") as z:
    print("Contenido de textos.zip:")
    print(z.namelist())
```
